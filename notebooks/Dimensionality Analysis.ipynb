```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Analisi Embeddings degli Writer tramite PCA
Visualizzazione della separabilità degli writer nello spazio embedding
per modelli addestrati con Binary Cross-Entropy Loss
"""

import torch
import torch.nn as nn
from torchvision import models, transforms
from torch.utils.data import Dataset, DataLoader
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import pandas as pd
from typing import Dict, List, Tuple
import warnings
warnings.filterwarnings('ignore')

# Configurazione stile per pubblicazione
plt.style.use('seaborn-v0_8-paper')
sns.set_palette("husl")
plt.rcParams['figure.dpi'] = 300
plt.rcParams['font.size'] = 10
plt.rcParams['axes.labelsize'] = 11
plt.rcParams['axes.titlesize'] = 12

# ============================================================================
# DEFINIZIONE ARCHITETTURA MODELLO
# ============================================================================

class BaseSiameseNetwork(nn.Module):
    """Base class for Siamese networks"""
    
    def __init__(self, encoder: nn.Module, feature_dim: int, projection_dim: int = 512):
        super().__init__()
        self.encoder = encoder
        self.feature_dim = feature_dim
        self.projection_dim = projection_dim
        self.fc = self._build_fc_block(feature_dim * 2, projection_dim)
    
    def _build_fc_block(self, input_dim: int, projection_dim: int) -> nn.Sequential:
        half = projection_dim // 2
        quarter = half // 2
        
        return nn.Sequential(
            nn.Linear(input_dim, projection_dim),
            nn.BatchNorm1d(projection_dim),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(projection_dim, half),
            nn.BatchNorm1d(half),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(half, quarter),
            nn.BatchNorm1d(quarter),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(quarter, 1),
            nn.Sigmoid()
        )
    
    def forward_one(self, x: torch.Tensor) -> torch.Tensor:
        x = self.encoder(x)
        return x.view(x.size(0), -1)
    
    def get_embedding(self, x: torch.Tensor) -> torch.Tensor:
        """Extract normalized embeddings for an image"""
        feat = self.forward_one(x)
        return nn.functional.normalize(feat, p=2, dim=1)

class SiameseMobileNetV3Small(BaseSiameseNetwork):
    """Siamese network based on MobileNetV3-Small"""
    
    def __init__(self, in_channels: int = 1, projection_dim: int = 512):
        mobilenet = models.mobilenet_v3_small(weights=None)
        mobilenet.features[0][0] = nn.Conv2d(in_channels, 16, kernel_size=3, stride=2, padding=1, bias=False)
        encoder = nn.Sequential(mobilenet.features, nn.AdaptiveAvgPool2d(1))
        super().__init__(encoder=encoder, feature_dim=576, projection_dim=projection_dim)

# ============================================================================
# DATASET E CARICAMENTO DATI
# ============================================================================

class WriterImageDataset(Dataset):
    """Dataset for loading all images from specific writers"""
    
    def __init__(self, writer_paths: List[Path], writer_ids: List[str], 
                 dataset_names: List[str], transform=None, max_images_per_writer: int = 50):
        """
        Args:
            writer_paths: List of paths to writer directories
            writer_ids: List of writer IDs
            dataset_names: List of dataset names (IAM or RIMES)
            transform: Image transformations
            max_images_per_writer: Maximum number of images to load per writer
        """
        self.transform = transform
        self.images = []
        self.labels = []
        self.writer_ids = []
        self.dataset_names = []
        
        for writer_path, writer_id, dataset_name in zip(writer_paths, writer_ids, dataset_names):
            # Get all PNG images from writer directory
            image_files = sorted(list(writer_path.glob("*.png")))[:max_images_per_writer]
            
            for img_file in image_files:
                self.images.append(img_file)
                self.labels.append(writer_id)
                self.writer_ids.append(writer_id)
                self.dataset_names.append(dataset_name)
        
        print(f"Loaded {len(self.images)} images from {len(set(self.labels))} writers")
    
    def __len__(self):
        return len(self.images)
    
    def __getitem__(self, idx):
        img_path = self.images[idx]
        image = Image.open(img_path).convert('L')
        
        if self.transform:
            image = self.transform(image)
        
        return image, self.labels[idx], self.dataset_names[idx]

def get_writer_samples(base_path: Path, dataset_name: str, num_writers: int = 5) -> Tuple[List[Path], List[str]]:
    """
    Get paths and IDs for writers with most data from a dataset
    
    Args:
        base_path: Base path to datasets
        dataset_name: 'iam' or 'rimes'
        num_writers: Number of writers to sample (will select top N with most images)
    
    Returns:
        Tuple of (writer_paths, writer_ids)
    """
    if dataset_name.lower() == 'iam':
        dataset_path = base_path / 'iam_processed'
    else:
        dataset_path = base_path / 'rimes_processed'
    
    # Get all writer directories with their image counts
    writer_info = []
    for writer_dir in dataset_path.iterdir():
        if writer_dir.is_dir():
            # Count PNG files in this directory
            num_images = len(list(writer_dir.glob("*.png")))
            writer_info.append({
                'path': writer_dir,
                'name': writer_dir.name,
                'num_images': num_images
            })
    
    # Sort by number of images (descending) and take top N
    writer_info_sorted = sorted(writer_info, key=lambda x: x['num_images'], reverse=True)
    top_writers = writer_info_sorted[:num_writers]
    
    writer_paths = [w['path'] for w in top_writers]
    writer_ids = [f"{dataset_name.upper()}_{w['name']}" for w in top_writers]
    
    # Print selection info
    print(f"  {dataset_name.upper()}: Selected top {num_writers} writers with most images")
    for w in top_writers:
        print(f"    - {w['name']}: {w['num_images']} images")
    
    return writer_paths, writer_ids

# ============================================================================
# ESTRAZIONE EMBEDDINGS
# ============================================================================

def load_model(checkpoint_path: Path, device: torch.device) -> nn.Module:
    """Load trained model from checkpoint"""
    model = SiameseMobileNetV3Small(in_channels=1, projection_dim=128)
    
    # Load checkpoint
    checkpoint = torch.load(checkpoint_path, map_location=device)
    
    # Handle different checkpoint formats
    if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
        # Checkpoint saved as dict with metadata
        model.load_state_dict(checkpoint['model_state_dict'])
        epoch_info = f"Trained for {checkpoint.get('epoch', 'N/A')} epochs"
    else:
        # Checkpoint saved as state_dict directly
        model.load_state_dict(checkpoint)
        epoch_info = "Epoch info not available"
    
    model.to(device)
    model.eval()
    
    print(f"✓ Model loaded from {checkpoint_path.name}")
    print(f"  {epoch_info}")
    
    return model

def extract_embeddings(model: nn.Module, dataloader: DataLoader, 
                       device: torch.device) -> Tuple[np.ndarray, List[str], List[str]]:
    """
    Extract embeddings for all images
    
    Returns:
        embeddings: numpy array of shape (N, embedding_dim)
        labels: list of writer IDs
        datasets: list of dataset names
    """
    embeddings_list = []
    labels_list = []
    datasets_list = []
    
    print("Extracting embeddings...")
    with torch.no_grad():
        for batch_idx, (images, labels, datasets) in enumerate(dataloader):
            images = images.to(device)
            
            # Get embeddings
            emb = model.get_embedding(images)
            
            embeddings_list.append(emb.cpu().numpy())
            labels_list.extend(labels)
            datasets_list.extend(datasets)
            
            if (batch_idx + 1) % 10 == 0:
                print(f"  Processed {(batch_idx + 1) * dataloader.batch_size} images")
    
    embeddings = np.vstack(embeddings_list)
    print(f"✓ Extracted {embeddings.shape[0]} embeddings of dimension {embeddings.shape[1]}")
    
    return embeddings, labels_list, datasets_list

# ============================================================================
# VISUALIZZAZIONE CON PCA E t-SNE
# ============================================================================

def plot_pca_embeddings(embeddings: np.ndarray, labels: List[str], 
                       datasets: List[str], title: str, save_path: Path = None):
    """
    Plot 2D PCA projection of embeddings
    """
    print("Computing PCA...")
    pca = PCA(n_components=2)
    embeddings_2d = pca.fit_transform(embeddings)
    
    explained_var = pca.explained_variance_ratio_
    print(f"  PC1 explains {explained_var[0]*100:.2f}% variance")
    print(f"  PC2 explains {explained_var[1]*100:.2f}% variance")
    print(f"  Total explained: {sum(explained_var)*100:.2f}%")
    
    # Create DataFrame
    df = pd.DataFrame({
        'PC1': embeddings_2d[:, 0],
        'PC2': embeddings_2d[:, 1],
        'Writer': labels,
        'Dataset': datasets
    })
    
    # Plot
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # Get unique writers and assign colors
    unique_writers = sorted(df['Writer'].unique())
    colors = plt.cm.tab20(np.linspace(0, 1, len(unique_writers)))
    color_map = dict(zip(unique_writers, colors))
    
    # Plot each writer separately
    for writer in unique_writers:
        writer_data = df[df['Writer'] == writer]
        dataset = writer_data['Dataset'].iloc[0]
        
        # Different markers for different datasets
        marker = 'o' if 'IAM' in dataset else 's'
        
        ax.scatter(writer_data['PC1'], writer_data['PC2'], 
                  c=[color_map[writer]], marker=marker, s=100, 
                  alpha=0.6, edgecolors='black', linewidth=0.5,
                  label=writer)
    
    # Calculate centroids for each writer
    for writer in unique_writers:
        writer_data = df[df['Writer'] == writer]
        centroid_x = writer_data['PC1'].mean()
        centroid_y = writer_data['PC2'].mean()
        
        # Plot centroid
        ax.scatter(centroid_x, centroid_y, c=[color_map[writer]], 
                  marker='*', s=500, edgecolors='black', linewidth=2,
                  zorder=10)
        
        # Add writer label at centroid
        ax.annotate(writer, (centroid_x, centroid_y), 
                   fontsize=8, fontweight='bold',
                   ha='center', va='bottom',
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white', 
                            edgecolor='black', alpha=0.7))
    
    ax.set_xlabel(f'PC1 ({explained_var[0]*100:.1f}% variance)')
    ax.set_ylabel(f'PC2 ({explained_var[1]*100:.1f}% variance)')
    ax.set_title(f'{title}\nPCA Projection of Writer Embeddings')
    
    # Custom legend
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor='gray', 
               markersize=10, label='IAM Dataset', markeredgecolor='black'),
        Line2D([0], [0], marker='s', color='w', markerfacecolor='gray', 
               markersize=10, label='RIMES Dataset', markeredgecolor='black'),
        Line2D([0], [0], marker='*', color='w', markerfacecolor='gold', 
               markersize=15, label='Writer Centroid', markeredgecolor='black')
    ]
    ax.legend(handles=legend_elements, loc='best', frameon=True, fancybox=True)
    
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, bbox_inches='tight', dpi=300)
    plt.show()
    
    return pca, embeddings_2d, df

def plot_tsne_embeddings(embeddings: np.ndarray, labels: List[str], 
                        datasets: List[str], title: str, save_path: Path = None):
    """
    Plot 2D t-SNE projection of embeddings
    """
    print("Computing t-SNE (this may take a while)...")
    tsne = TSNE(n_components=2, random_state=42, perplexity=30, max_iter=1000)
    embeddings_2d = tsne.fit_transform(embeddings)
    
    # Create DataFrame
    df = pd.DataFrame({
        'Dim1': embeddings_2d[:, 0],
        'Dim2': embeddings_2d[:, 1],
        'Writer': labels,
        'Dataset': datasets
    })
    
    # Plot
    fig, ax = plt.subplots(figsize=(12, 10))
    
    unique_writers = sorted(df['Writer'].unique())
    colors = plt.cm.tab20(np.linspace(0, 1, len(unique_writers)))
    color_map = dict(zip(unique_writers, colors))
    
    for writer in unique_writers:
        writer_data = df[df['Writer'] == writer]
        dataset = writer_data['Dataset'].iloc[0]
        marker = 'o' if 'IAM' in dataset else 's'
        
        ax.scatter(writer_data['Dim1'], writer_data['Dim2'], 
                  c=[color_map[writer]], marker=marker, s=100, 
                  alpha=0.6, edgecolors='black', linewidth=0.5,
                  label=writer)
    
    # Centroids
    for writer in unique_writers:
        writer_data = df[df['Writer'] == writer]
        centroid_x = writer_data['Dim1'].mean()
        centroid_y = writer_data['Dim2'].mean()
        
        ax.scatter(centroid_x, centroid_y, c=[color_map[writer]], 
                  marker='*', s=500, edgecolors='black', linewidth=2, zorder=10)
        
        ax.annotate(writer, (centroid_x, centroid_y), 
                   fontsize=8, fontweight='bold',
                   ha='center', va='bottom',
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white', 
                            edgecolor='black', alpha=0.7))
    
    ax.set_xlabel('t-SNE Dimension 1')
    ax.set_ylabel('t-SNE Dimension 2')
    ax.set_title(f'{title}\nt-SNE Projection of Writer Embeddings')
    
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor='gray', 
               markersize=10, label='IAM Dataset', markeredgecolor='black'),
        Line2D([0], [0], marker='s', color='w', markerfacecolor='gray', 
               markersize=10, label='RIMES Dataset', markeredgecolor='black'),
        Line2D([0], [0], marker='*', color='w', markerfacecolor='gold', 
               markersize=15, label='Writer Centroid', markeredgecolor='black')
    ]
    ax.legend(handles=legend_elements, loc='best', frameon=True, fancybox=True)
    
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, bbox_inches='tight', dpi=300)
    plt.show()
    
    return embeddings_2d, df

# ============================================================================
# ANALISI QUANTITATIVA SEPARABILITÀ
# ============================================================================

def compute_separation_metrics(embeddings: np.ndarray, labels: List[str]) -> pd.DataFrame:
    """
    Compute quantitative metrics for writer separation
    """
    from scipy.spatial.distance import pdist, squareform
    
    unique_writers = sorted(set(labels))
    metrics = []
    
    # Compute pairwise distances
    distances = squareform(pdist(embeddings, metric='euclidean'))
    
    for writer in unique_writers:
        writer_indices = [i for i, l in enumerate(labels) if l == writer]
        other_indices = [i for i, l in enumerate(labels) if l != writer]
        
        # Intra-writer distances (cohesion)
        if len(writer_indices) > 1:
            intra_dists = []
            for i in range(len(writer_indices)):
                for j in range(i+1, len(writer_indices)):
                    intra_dists.append(distances[writer_indices[i], writer_indices[j]])
            intra_mean = np.mean(intra_dists)
            intra_std = np.std(intra_dists)
        else:
            intra_mean = 0
            intra_std = 0
        
        # Inter-writer distances (separation)
        inter_dists = []
        for i in writer_indices:
            for j in other_indices:
                inter_dists.append(distances[i, j])
        inter_mean = np.mean(inter_dists)
        inter_std = np.std(inter_dists)
        
        # Separation ratio (higher is better)
        separation_ratio = inter_mean / (intra_mean + 1e-8)
        
        metrics.append({
            'Writer': writer,
            'Num_Samples': len(writer_indices),
            'Intra_Mean': intra_mean,
            'Intra_Std': intra_std,
            'Inter_Mean': inter_mean,
            'Inter_Std': inter_std,
            'Separation_Ratio': separation_ratio
        })
    
    df_metrics = pd.DataFrame(metrics)
    return df_metrics

def plot_separation_heatmap(embeddings: np.ndarray, labels: List[str], 
                           title: str, save_path: Path = None):
    """
    Plot heatmap of average distances between writer centroids
    """
    from scipy.spatial.distance import euclidean
    
    unique_writers = sorted(set(labels))
    n_writers = len(unique_writers)
    
    # Compute centroids
    centroids = {}
    for writer in unique_writers:
        writer_indices = [i for i, l in enumerate(labels) if l == writer]
        centroids[writer] = embeddings[writer_indices].mean(axis=0)
    
    # Compute pairwise distances between centroids
    dist_matrix = np.zeros((n_writers, n_writers))
    for i, writer1 in enumerate(unique_writers):
        for j, writer2 in enumerate(unique_writers):
            dist_matrix[i, j] = euclidean(centroids[writer1], centroids[writer2])
    
    # Plot heatmap
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(dist_matrix, annot=True, fmt='.2f', cmap='RdYlGn',
                xticklabels=unique_writers, yticklabels=unique_writers,
                cbar_kws={'label': 'Euclidean Distance'}, ax=ax)
    
    ax.set_title(f'{title}\nPairwise Distances Between Writer Centroids')
    ax.set_xlabel('Writer ID')
    ax.set_ylabel('Writer ID')
    
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, bbox_inches='tight', dpi=300)
    plt.show()

# ============================================================================
# PIPELINE COMPLETA
# ============================================================================

def analyze_experiment(exp_name: str, checkpoint_path: Path, 
                      dataset_base_path: Path, test_dataset: str,
                      output_dir: Path, device: torch.device):
    """
    Analyze embeddings for a single experiment
    
    Args:
        exp_name: Name of experiment (e.g., "IAM→IAM")
        checkpoint_path: Path to best.pth checkpoint
        dataset_base_path: Base path to datasets
        test_dataset: Dataset to test on ('iam' or 'rimes')
        output_dir: Directory to save outputs
        device: Device for computation
    """
    print("\n" + "="*80)
    print(f"ANALYZING: {exp_name}")
    print("="*80)
    
    # Create output directory
    exp_output = output_dir / exp_name.replace('→', '_to_')
    exp_output.mkdir(exist_ok=True, parents=True)
    
    # Load model
    model = load_model(checkpoint_path, device)
    
    # Get writer samples (5 from each dataset)
    print("\nSelecting writers...")
    iam_paths, iam_ids = get_writer_samples(dataset_base_path, 'iam', num_writers=5)
    rimes_paths, rimes_ids = get_writer_samples(dataset_base_path, 'rimes', num_writers=5)
    
    all_paths = iam_paths + rimes_paths
    all_ids = iam_ids + rimes_ids
    all_datasets = ['IAM']*5 + ['RIMES']*5
    
    print(f"Selected {len(all_ids)} writers: {all_ids}")
    
    # Create dataset and dataloader
    transform = transforms.Compose([
        transforms.Resize((128, 512)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5], std=[0.5])
    ])
    
    dataset = WriterImageDataset(all_paths, all_ids, all_datasets, 
                                 transform=transform, max_images_per_writer=30)
    dataloader = DataLoader(dataset, batch_size=32, shuffle=False, num_workers=4)
    
    # Extract embeddings
    embeddings, labels, datasets = extract_embeddings(model, dataloader, device)
    
    # Save embeddings
    np.save(exp_output / 'embeddings.npy', embeddings)
    with open(exp_output / 'labels.txt', 'w') as f:
        for label in labels:
            f.write(f"{label}\n")
    
    # PCA visualization
    print("\n" + "-"*80)
    pca, emb_2d_pca, df_pca = plot_pca_embeddings(
        embeddings, labels, datasets, 
        title=f"{exp_name}",
        save_path=exp_output / 'pca_projection.png'
    )
    df_pca.to_csv(exp_output / 'pca_coordinates.csv', index=False)
    
    # t-SNE visualization
    print("\n" + "-"*80)
    emb_2d_tsne, df_tsne = plot_tsne_embeddings(
        embeddings, labels, datasets,
        title=f"{exp_name}",
        save_path=exp_output / 'tsne_projection.png'
    )
    df_tsne.to_csv(exp_output / 'tsne_coordinates.csv', index=False)
    
    # Separation metrics
    print("\n" + "-"*80)
    print("Computing separation metrics...")
    sep_metrics = compute_separation_metrics(embeddings, labels)
    print("\nSeparation Metrics:")
    print(sep_metrics.to_string(index=False))
    sep_metrics.to_csv(exp_output / 'separation_metrics.csv', index=False)
    
    # Heatmap
    plot_separation_heatmap(embeddings, labels, 
                           title=f"{exp_name}",
                           save_path=exp_output / 'distance_heatmap.png')
    
    print(f"\n✓ Analysis complete for {exp_name}")
    print(f"  Results saved to: {exp_output}")

def analyze_all_experiments(experiments_base_path: Path, 
                           dataset_base_path: Path,
                           output_dir: Path = Path('./embedding_analysis')):
    """
    Analyze all experiments
    """
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")
    
    experiments = {
        'IAM→IAM': {
            'checkpoint': experiments_base_path / 'mobilenet_v3_small_iam_to_iam' / 'mobilenet_v3_small_iam_to_iam_best.pth',
            'test_dataset': 'iam'
        },
        'IAM→RIMES': {
            'checkpoint': experiments_base_path / 'mobilenet_v3_small_iam_to_rimes' / 'mobilenet_v3_small_iam_to_rimes_best.pth',
            'test_dataset': 'rimes'
        },
        'RIMES→IAM': {
            'checkpoint': experiments_base_path / 'mobilenet_v3_small_rimes_to_iam' / 'mobilenet_v3_small_rimes_to_iam_best.pth',
            'test_dataset': 'iam'
        },
        'RIMES→RIMES': {
            'checkpoint': experiments_base_path / 'mobilenet_v3_small_rimes_to_rimes' / 'mobilenet_v3_small_rimes_to_rimes_best.pth',
            'test_dataset': 'rimes'
        }
    }
    
    # Create main output directory
    output_dir.mkdir(exist_ok=True, parents=True)
    
    # Analyze each experiment
    for exp_name, exp_config in experiments.items():
        try:
            analyze_experiment(
                exp_name=exp_name,
                checkpoint_path=exp_config['checkpoint'],
                dataset_base_path=dataset_base_path,
                test_dataset=exp_config['test_dataset'],
                output_dir=output_dir,
                device=device
            )
        except Exception as e:
            print(f"\n❌ Error analyzing {exp_name}: {str(e)}")
            import traceback
            traceback.print_exc()
            continue
    
    print("\n" + "="*80)
    print("ALL ANALYSES COMPLETE")
    print(f"Results saved to: {output_dir.absolute()}")
    print("="*80)

# ============================================================================
# ESECUZIONE
# ============================================================================

if __name__ == "__main__":
    # CONFIGURAZIONE
    EXPERIMENTS_PATH = Path("../results/bce_experiments_20251220_234010")
    DATASET_PATH = Path("../datasets/processed-handwritten")
    OUTPUT_PATH = Path("./embedding_analysis_results")
    
    # Esegui analisi completa
    analyze_all_experiments(EXPERIMENTS_PATH, DATASET_PATH, OUTPUT_PATH)
    
    # Per analisi singola:
    # device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    # analyze_experiment(
    #     exp_name="IAM→IAM",
    #     checkpoint_path=EXPERIMENTS_PATH / "mobilenet_v3_small_iam_to_iam" / "mobilenet_v3_small_iam_to_iam_best.pth",
    #     dataset_base_path=DATASET_PATH,
    #     test_dataset='iam',
    #     output_dir=OUTPUT_PATH,
    #     device=device
    # )
```
